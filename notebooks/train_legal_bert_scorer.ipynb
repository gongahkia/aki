```python
from src.ml.legal_bert_scorer import LegalBertScorer
from src.ml.data import load_data

data = load_data('data/generated/ml_bootstrap_labels.csv')
train_df = data['train']
test_df = data['test']
print(f'train={len(train_df)} test={len(test_df)}')
```

```python
scorer = LegalBertScorer()
metrics = scorer.train(
    train_df['text'].tolist(),
    train_df['quality_score'].tolist(),
    num_epochs=8,
    batch_size=8,
)
print(metrics)
```

```python
eval_metrics = scorer.evaluate(
    test_df['text'].tolist(),
    test_df['quality_score'].tolist(),
)
print(eval_metrics)
```

```python
scorer.save_model('models/legal_bert_scorer')
print('Saved to models/legal_bert_scorer')
```
